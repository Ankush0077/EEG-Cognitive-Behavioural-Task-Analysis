```python
# Re-import needed modules due to code execution state reset
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import scipy.stats as stats

# Re-load the datasets
final_prl_data = pd.read_csv('data/beads_data_with_conf_slope.csv')
task_df1 = pd.read_csv('data/Schizotypy.csv')

# Rename participant_id to PID for merge
final_prl_data = final_prl_data.rename(columns={'participant_id': 'PID'})
merged_df = pd.merge(final_prl_data, task_df1, on='PID')

# Select only numeric columns
numeric_cols = merged_df.select_dtypes(include=['float64', 'int64'])

# Function to compute correlation and p-values
def calculate_correlations_with_pvalues(df):
    cols = df.columns
    corr_matrix = pd.DataFrame(index=cols, columns=cols, dtype=float)
    pval_matrix = pd.DataFrame(index=cols, columns=cols, dtype=float)
    for col1 in cols:
        for col2 in cols:
            if col1 == col2:
                corr_matrix.loc[col1, col2] = 1.0
                pval_matrix.loc[col1, col2] = 0.0
            else:
                corr, pval = stats.pearsonr(df[col1], df[col2])
                corr_matrix.loc[col1, col2] = corr
                pval_matrix.loc[col1, col2] = pval
    return corr_matrix, pval_matrix

# Compute matrices
corr_matrix, pval_matrix = calculate_correlations_with_pvalues(numeric_cols)

# Filter: keep correlations where p-value < 0.1
significant_corrs = (pval_matrix < 0.2) & (pval_matrix > 0)
filtered_corrs = corr_matrix.where(significant_corrs)

# Clean up for heatmap (drop rows/cols with all NaNs)
filtered_corrs_cleaned = filtered_corrs.dropna(axis=0, how='all').dropna(axis=1, how='all')

# Plot the filtered heatmap
plt.figure(figsize=(16, 12))
sns.heatmap(filtered_corrs_cleaned, annot=True, cmap='coolwarm', fmt='.2f', square=True)
plt.title('Significant Correlations (p < 0.1)', fontsize=18)
plt.show()
```
